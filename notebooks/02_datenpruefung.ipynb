```python
import numpy as np
import pandas as pd

parameter = pd.DataFrame(
    {
        "Signalvarianz": [2.32, 2.66, 0.21],
        "V": [14.1, 14.2, 2.95],
        "N": [34, 17, 17],
    },
    index=["X: Vorsaison-Chancen", "Y: Hinrunden-Chancen", "Z: Rückrunden-Tore"],
)

parameter["Zufallsvarianz"] = parameter["V"] / parameter["N"]

parameter["Gesamtvarianz"] = (
    parameter["Signalvarianz"] + parameter["Zufallsvarianz"]
)

parameter["Signalstandardabweichung"] = np.sqrt(
    parameter["Signalvarianz"]
)

parameter["Zufallsstandardabweichung"] = np.sqrt(
    parameter["Zufallsvarianz"]
)

parameter.round(3)
```
